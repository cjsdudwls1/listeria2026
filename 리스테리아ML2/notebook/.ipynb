```python
# NOTE: 이 데이터셋에는 'date_basis'(기준일)과 'addr_source'(주소 출처)가 보이므로,
# "좌표(lat, lon) + 기준일(date_basis) + 품목코드(prdlst_code) + 주소출처(addr_source)" 기준으로
# 같은 행을 찾습니다. 다른 기준을 원하면 key_cols만 바꾸면 됩니다.
key_cols = ["lat", "lon", "date_basis", "prdlst_code", "addr_source"]

dup_rows = df[df.duplicated(subset=key_cols, keep=False)].sort_values(key_cols).reset_index(drop=True)

dup_rows
```
