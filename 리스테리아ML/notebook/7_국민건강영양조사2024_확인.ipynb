- 컬럼 목록·선정 여부: `리스테리아ML/docs/02_데이터수집/국민건강영양조사2024_컬럼목록.xlsx`

```python
from pathlib import Path
import pandas as pd
pd.set_option('display.max_columns', 200)
pd.set_option('display.max_rows', 200)
pd.set_option('display.max_colwidth', 80)

DATA = Path('../data/collected/knhanes_2024')
XLSX = Path('../docs/02_데이터수집/국민건강영양조사2024_컬럼목록.xlsx')
for f in sorted(DATA.glob('*.parquet')):
    print(f.name, f'{f.stat().st_size/1e6:.1f} MB')
```

## 1. `hn24_all.parquet` (1명 1행)

```python
df_all = pd.read_parquet(DATA / 'hn24_all.parquet')
print(df_all.shape)
df_all.head()
```

## 2. `hn24_24rc.parquet` (재료 1개 1행)

```python
df_rc = pd.read_parquet(DATA / 'hn24_24rc.parquet')
print(df_rc.shape, '/ 사람 수(ID)', df_rc['ID'].nunique() if 'ID' in df_rc else '-')
df_rc.head(20)
```

```python
df_rc
```

## 3. 컬럼별 자료형·결측·고유값 수

```python
def col_info(df):
    return pd.DataFrame({'dtype': df.dtypes.astype(str), '결측수': df.isna().sum(),
                         '결측률(%)': (df.isna().mean() * 100).round(2), '고유값수': df.nunique(),
                         '예시값': df.iloc[0]})

col_info(df_rc)
```

```python
col_info(df_all)
```

## 4. 컬럼 목록 엑셀 (`국민건강영양조사2024_컬럼목록.xlsx`) — 시트별

```python
sheets = pd.read_excel(XLSX, sheet_name=None)
for name, s in sheets.items():
    print(f'[{name}]', s.shape)
    display(s)
```
